```python
import pandas as pd
import os

print(os.listdir("/content"))
```

```python
df = pd.read_csv("/content/train.csv")

print(df.head())
print("Dataset shape:", df.shape)
```

```python

# Import required libraries
import pandas as pd
import numpy as np

# Display dataset dimensions
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display data types and non-null values
print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all").T)
```

```python

# Check missing values before cleaning
print("Missing values before cleaning:")
print(df[["Age", "Cabin", "Embarked"]].isnull().sum())

# Store the original median and mode for documentation
age_median = df["Age"].median()
embarked_mode = df["Embarked"].mode()[0]

print("\nAge median:", age_median)
print("Embarked mode:", embarked_mode)

# Handle missing values
df["Age"] = df["Age"].fillna(age_median)
df["Embarked"] = df["Embarked"].fillna(embarked_mode)
df["Cabin"] = df["Cabin"].fillna("Unknown")

# Verify the result
print("\nMissing values after cleaning:")
print(df[["Age", "Cabin", "Embarked"]].isnull().sum())

print("\nUpdated dataset shape:", df.shape)
```

```python

# 1. Check complete duplicate rows
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)

# 2. Check duplicate Passenger IDs
duplicate_ids = df["PassengerId"].duplicated().sum()
print("Number of duplicate Passenger IDs:", duplicate_ids)

# 3. Check unique values in categorical columns
print("\nUnique values in Sex:")
print(df["Sex"].unique())

print("\nUnique values in Embarked:")
print(df["Embarked"].unique())

# 4. Check invalid values in important columns
print("\nInvalid Sex values:")
print(df[~df["Sex"].isin(["male", "female"])]["Sex"])

print("\nInvalid Survived values:")
print(df[~df["Survived"].isin([0, 1])]["Survived"])

print("\nInvalid Pclass values:")
print(df[~df["Pclass"].isin([1, 2, 3])]["Pclass"])
```

```python

# Detect potential outliers using the IQR method

numeric_columns = ["Age", "Fare", "SibSp", "Parch"]

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\nColumn: {column}")
    print("Q1:", round(Q1, 2))
    print("Q3:", round(Q3, 2))
    print("IQR:", round(IQR, 2))
    print("Lower bound:", round(lower_bound, 2))
    print("Upper bound:", round(upper_bound, 2))
    print("Potential outliers:", len(outliers))
```

```python

# Final validation of the cleaned dataset

print("Final dataset shape:", df.shape)

print("\nMissing values in the dataset:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

# Preview cleaned data
display(df.head())

# Save the cleaned dataset
df.to_csv("titanic_cleaned.csv", index=False)

print("\nCleaned dataset saved as titanic_cleaned.csv")
```
